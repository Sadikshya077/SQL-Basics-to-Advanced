```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
query = '''
    select 
    sc.customer_id, sc.first_name, sc.last_name, sc.email, sc.street, so.order_status, so.order_date
    from sales.customers sc
    join sales.orders so
    on sc.customer_id = so.customer_id
    where sc.state= 'TX' and so.order_status=3;
'''
query_df =pd.read_sql(query,engine)
```
